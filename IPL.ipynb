```python
#19
tossdata = matches.groupby(['season', 'toss_decision']).size().unstack()
tossdata
tossdata.plot(kind='bar', figsize=(20,10))
plt.title("Toss Decisions Across Seasons")
plt.xlabel("Season")
plt.ylabel("Number of Toss Decisions")
plt.show()
```

```python
#21
plt.figure(figsize=(20, 10))
matches['winner'].value_counts().plot(kind='bar')
plt.title('Distribution of Match Winners')
plt.xlabel('Teams')
plt.ylabel('Number of Wins')
plt.xticks(rotation=45, ha='right')

plt.show()
```

```python
#22
tossoutcomes = matches.groupby(['toss_winner', 'toss_decision']).size().unstack()

plt.figure(figsize=(15,5))
tossoutcomes.plot(kind='bar', stacked=True)
plt.title('Toss Outcomes for All Teams')
plt.xlabel('Teams')
plt.ylabel('Number of Tosses Won')
plt.xticks(rotation=45,ha='right')
plt.show()
```

```python
#23
wins = matches['winner'].value_counts()
top5 = wins.head(5)
plt.figure(figsize=(10,6))
top5.plot(kind='bar')
plt.title("Top 5 Teams with Most Wins (All Seasons)")
plt.xlabel("Teams")
plt.ylabel("Total Wins")
plt.xticks(rotation=45)
plt.show()
```
