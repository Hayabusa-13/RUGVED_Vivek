# Linear Regression


```python
#y = mx+c

x = co2['Fuel Consumption Comb (L/100 km)'].values
y = co2['CO2 Emissions(g/km)'].values
print(x)
print(y)
```

```python
xm = np.mean(x)
ym = np.mean(y)
print(xm)
print(ym)
```

```python
a = np.sum((x-xm)*(y-ym))
b= np.sum((x-xm)**2)
m= a/b
```

```python
c= ym- m*xm
```

```python
m,c
```

```python
yfin = m*x + c
```

```python
plt.scatter(x, y)
plt.plot(x, yfin)
plt.xlabel('Fuel Consumption Comb (L/100 km)')
plt.ylabel('CO2 Emissions (g/km)')
plt.title('LR of fuelconsump comb (l/100)km w.r.t to co2 emissions')
plt.show()
```
